## **Calculate fingerprint descriptors**


### **Calculate PaDEL descriptors**

```python
import subprocess

try:
    java_version = subprocess.check_output(['java', '-version'], stderr=subprocess.STDOUT, text=True)
    print(java_version)
except FileNotFoundError:
    print("Java is not found. You may need to add it to the PATH.")

```

```python
from padelpy import padeldescriptor

# Specify the input SMILES file and output CSV file
padeldescriptor(
    mol_dir='molecules.smi',  # SMILES file path
    d_file='descriptors.csv',  # Output file for descriptors
    fingerprints=True, # Set to True if you want fingerprints
)
```
